```python
# Group data by "Product_Name", sum "Quantity" , sort in descending order and display the top product.

best_selling_product = sales.groupby("Product_Name")["Quantity"].sum().sort_values(ascending = False).head(1)
best_selling_product
```

```python
# Group data by "Category" , sum "Total_Bill_Amount" , sort in descending order and display the top product.

category_sales = sales.groupby("Category")["Total_Bill_Amount"].sum().sort_values(ascending = False).head(1)
category_sales
```

```python
# Group data by "Category" , sum "Total_Bill_Amount" and sort in descending order.

category_sales = sales.groupby("Category")["Total_Bill_Amount"].sum().sort_values(ascending = False)
category_sales
```

```python
# Plot a bar graph for category-wise sales.

plt.style.use("grayscale")

plt.figure(figsize = (12,6))

plt.bar(category_sales.index, category_sales.values, color = "pink")
plt.xlabel("Category")
plt.ylabel("Category_Sales")
plt.title("Sales Based On Category")

plt.show()
```

```python
# Group data by "Store_Location" , sum "Total_Bill_Amount" and sort in descending order.

city_sales = sales.groupby("Store_Location")["Total_Bill_Amount"].sum().sort_values(ascending = False)
city_sales
```

```python
# Plot a pie chart for city-wise sales.

plt.style.use("seaborn-v0_8-deep")

plt.figure(figsize = (13,6))

plt.pie(city_sales.values, labels = city_sales.index, colors = ["yellow", "beige", "grey", "purple", "red"], autopct = "%0.1f%%", shadow = True)
plt.title("City Wise Sales")

plt.show()
```

```python
# Group data by "Product_Name" , sum "Total_Bill_Amount" , sort in descending order and display the first five rows.

top_5_product = sales.groupby("Product_Name")["Total_Bill_Amount"].sum().sort_values(ascending = False).head(5)
top_5_product
```

```python
# Plot a bar graph for top 5 products.

plt.style.use("seaborn-v0_8-whitegrid")
plt.figure(figsize = (15, 6))

plt.bar(top_5_product.index, top_5_product.values, color = "orange")
plt.xlabel("Products")
plt.ylabel("Revenue")
plt.title("Top 5 Products")
plt.xticks(rotation = 25)

plt.show()
```

```python
# Group data by "Category" , sum "Quantity" and  sort in descending order.

category_quantity = sales.groupby("Category")["Quantity"].sum().sort_values(ascending = False)
category_quantity
```

```python
# Plot a bar graph for category-wise quantity.

plt.style.use("classic")
plt.figure(figsize = (14, 6))

plt.bar(category_quantity.index, category_quantity.values)
plt.xlabel("Category")
plt.ylabel("Quantity")
plt.title("Category-Wise Total Quantity Sold")
plt.xticks(rotation = 25)

plt.show()
```

```python
# Group data by "Product_Name" , sum "Quantity" , sort in descending order and display first five rows.

top_5_quantity_product = sales.groupby("Product_Name")["Quantity"].sum().sort_values(ascending = False).head(5)
top_5_quantity_product
```

```python
# Group data by "Category" , average "Price_Per_Unit" and sort in descending order.

avg_price_category = sales.groupby("Category")["Price_Per_Unit"].mean().sort_values(ascending = False)
avg_price_category
```

```python
# Group data by "Product_Name" and aggerate both "Quantity" and "Total_Bill_Amount" sums.

product_analysis = sales.groupby("Product_Name").agg({"Quantity" : "sum", "Total_Bill_Amount" : "sum"})
product_analysis
```

```python
# Filter out high-values bills where the total amount is greater than 500.

high_value_bills = sales[sales["Total_Bill_Amount"] > 500]
high_value_bills.shape[0]
```

```python
# Display the average of high_value_bills.

print(high_value_bills["Total_Bill_Amount"].mean())
```

```python
# Display the maximum of high_value_bills.

print(high_value_bills["Total_Bill_Amount"].max())
```

```python
# Display the total sum of high_value_bills.

print(high_value_bills["Total_Bill_Amount"].sum())
```

```python
# Count the total number of unique bills generated at each store locations.

store_bill_count = sales.groupby("Store_Location")["Bill_ID"].nunique().sort_values(ascending = False)
store_bill_count
```

```python
# Plot a bar graph for store-wise bill count.

plt.style.use("seaborn-v0_8-deep")
plt.figure(figsize = (15,6))

plt.bar(store_bill_count.index, store_bill_count.values)
plt.xlabel("City")
plt.ylabel("Bill Count")
plt.title("Store-Wise Bill Count")

plt.show()
```

```python
# Calculate total revenue percent for each category.

category_revenue = sales.groupby("Category")["Total_Bill_Amount"].sum()

total_revenue = sales["Total_Bill_Amount"].sum()

category_revenue_percent = (category_revenue / total_revenue) * 100

category_revenue_percent
```
